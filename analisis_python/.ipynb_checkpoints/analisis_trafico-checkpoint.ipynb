## Sección 2 — Análisis Exploratorio

```python
print('=== Distribución de protocolos ===')
conteo_proto = df['protocolo'].value_counts()
print(conteo_proto.to_string())
print(f'\nPorcentaje TCP: {conteo_proto.get("TCP", 0) / len(df) * 100:.1f}%')
print(f'Porcentaje UDP: {conteo_proto.get("UDP", 0) / len(df) * 100:.1f}%')
```

```python
print('=== Top 10 puertos destino más usados ===')
print(df['puerto_destino'].value_counts().head(10).to_string())
```

```python
print('=== Equipos (IPs origen) más activos ===')
activos = df.groupby('ip_origen').agg(
    paquetes=('longitud', 'count'),
    bytes_total=('longitud', 'sum')
).sort_values('paquetes', ascending=False)
print(activos.head(10).to_string())
```

```python
print('=== Categorías de servicio detectadas ===')
print(df['servicio'].value_counts().to_string())
```

```python
print('=== Tráfico por hora del día ===')
print(df.groupby('hora').size().rename('paquetes').to_string())
```

## Sección 3 — Modelos de Data Science

### 3.1 Clasificación KNN

El modelo KNN clasifica el **tipo de tráfico** (servicio de red) de cada paquete
usando como características: longitud del paquete, puerto origen, puerto destino y protocolo.
La etiqueta de clasificación es la **categoría de servicio** derivada del puerto destino.

```python
from modelos import clasificar_tipo_trafico

res_knn = clasificar_tipo_trafico(df, k=5)
```

### 3.2 Clustering K-Means

K-Means agrupa los **equipos** (IPs de origen) según su comportamiento de red:
volumen de tráfico total, bytes enviados, puertos únicos utilizados y porcentaje TCP.
Se usan 3 clusters: **Intensivo**, **Moderado** y **Ligero**.

```python
from modelos import clustering_equipos

res_cluster = clustering_equipos(df, k=3)
```

### 3.3 Análisis Predictivo

Modelo de **regresión logística** que predice el protocolo predominante (TCP/UDP)
en función de la hora del día. Permite estimar qué protocolo es más probable
en un horario determinado.

```python
from modelos import predecir_protocolo_por_hora

res_pred = predecir_protocolo_por_hora(df)
```

## Sección 5 — Conclusiones

> Las celdas siguientes generan automáticamente un resumen cuantitativo.
> Al final puedes agregar tu interpretación personal.

```python
total = len(df)
tcp_pct = df['protocolo'].value_counts(normalize=True).get('TCP', 0) * 100
udp_pct = 100 - tcp_pct
puerto_top = int(df['puerto_destino'].value_counts().idxmax())
ip_mas_activa = df['ip_origen'].value_counts().idxmax()

print('=' * 60)
print('  RESUMEN DEL ANÁLISIS')
print('=' * 60)
print(f'  Total de paquetes analizados   : {total:,}')
print(f'  Protocolo TCP                  : {tcp_pct:.1f}%')
print(f'  Protocolo UDP                  : {udp_pct:.1f}%')
print(f'  Puerto destino más usado       : {puerto_top}')
print(f'  IP más activa (origen)         : {ip_mas_activa}')
print(f'  Accuracy KNN                   : {res_knn["accuracy"]:.2%}')
print(f'  Silhouette score K-Means       : {res_cluster["silhouette"]:.4f}')
print(f'  Accuracy predicción protocolo  : {res_pred["accuracy"]:.2%}')
print('=' * 60)
print()
print('Distribución de equipos por nivel de uso:')
print(res_cluster['datos_equipos']['nivel_uso'].value_counts().to_string())
```

### Interpretación de resultados

*(Completa esta sección con tus observaciones después de ejecutar el notebook)*

**Protocolo predominante:**  
El tráfico analizado muestra una clara predominancia de TCP sobre UDP, lo que es consistente con el uso típico de navegación web (HTTPS puerto 443) y transferencia de datos.

**Puertos más utilizados:**  
El puerto 443 (HTTPS) es el más frecuente, lo que indica que la mayor parte del tráfico corresponde a navegación web segura. El puerto 53 (DNS) también aparece de forma significativa.

**Equipos más activos:**  
La IP local principal genera la mayor cantidad de tráfico, lo que es esperado al ser el equipo capturando en modo promiscuo. Las IPs externas corresponden a servicios en la nube (Microsoft, Cloudflare, etc.).

**Modelo KNN:**  
El clasificador KNN logra identificar correctamente los tipos de tráfico basándose en los puertos y características del paquete. La precisión depende del volumen de datos.

**Clustering K-Means:**  
Los equipos se agrupan en tres niveles de uso claramente diferenciados. Los equipos intensivos generan mayor volumen de tráfico y utilizan más puertos variados.

**Análisis predictivo:**  
El modelo de regresión logística muestra que TCP domina en la mayoría de horas, con ligeras variaciones hacia UDP en horarios específicos (DNS, streaming).
